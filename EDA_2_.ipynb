# Identify numerical and categorical columns


```python
numerical_cols = ["age", "fnlwgt", "education_num", "capital_gain", "capital_loss", "hours_per_week"]
categorical_cols = ["workclass", "education", "marital_status", "occupation", "relationship", "race", "sex", "native_country", "income"]
```

# Scaling Numerical Features


```python
scaler_standard = StandardScaler()
df_standard_scaled = df.copy()
df_standard_scaled[numerical_cols] = scaler_standard.fit_transform(df[numerical_cols])
```

```python
scaler_minmax = MinMaxScaler()
df_minmax_scaled = df.copy()
df_minmax_scaled[numerical_cols] = scaler_minmax.fit_transform(df[numerical_cols])
```

# Encoding Categorical Variables

```python
one_hot_cols = ["race", "sex", "income"]
label_cols = ["workclass", "education", "marital_status", "occupation", "relationship", "native_country"]
```

```python
df_encoded = pd.get_dummies(df, columns=one_hot_cols, drop_first=True)
label_encoders = {col: LabelEncoder() for col in label_cols}
for col in label_cols:
    df_encoded[col] = label_encoders[col].fit_transform(df_encoded[col])
```

# Feature Engineering

```python
df_encoded['capital_gain_log'] = np.log1p(df_encoded['capital_gain'])  # Log transformation
df_encoded['capital_total'] = df_encoded['capital_gain'] - df_encoded['capital_loss']  # New feature
df_encoded['work_hours_category'] = pd.cut(df_encoded['hours_per_week'], bins=[0, 20, 40, 60, 100], labels=["Part-time", "Full-time", "Overtime", "Extreme"])
df_encoded['work_hours_category'] = LabelEncoder().fit_transform(df_encoded['work_hours_category'])
```

# Outlier Removal using Isolation Forest

```python
iso_forest = IsolationForest(contamination=0.05, random_state=42)
outliers = iso_forest.fit_predict(df_encoded[numerical_cols])
df_filtered = df_encoded[outliers == 1]
```

```python
income_col = [col for col in df_filtered.columns if "income" in col][0]
```

# Feature Selection using PPS Score

```python
print("Generating PPS matrix...")
df_filtered_clean = df_filtered.dropna()
pps_matrix = pps.matrix(df_filtered_clean)
```

# Debugging step - print the entire PPS matrix structure

```python
print("PPS Matrix Structure:")
print(pps_matrix.head())
print("PPS Matrix Columns:", list(pps_matrix.columns))
```

```python
if not pps_matrix.empty and 'ppscore' in pps_matrix.columns:
    pps_scores = pps_matrix[['x', 'y', 'ppscore']].pivot(index='x', columns='y', values='ppscore')
    print("\nTop PPS Scores:\n", pps_scores.sort_values(by=income_col, ascending=False)[income_col].dropna())
else:
    print("Error: 'ppscore' column not found in PPS matrix. Using correlation as a fallback.")
    correlation_matrix = df_filtered.corr()
    correlation_with_target = correlation_matrix[income_col].sort_values(ascending=False)
    print("\nCorrelation with Income:\n", correlation_with_target)
```
